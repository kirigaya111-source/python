```python
n = int(input("Nhập n: "))
tong = 0

for i in range(2, n + 1, 2):
    tong += i

print(tong)
```

```python
s = input("Nhập chuỗi: ")

words = s.split()
so_tu = len(words)
tu_dai_nhat = max(words, key=len)

print("Words:", so_tu)
print("Longest word:", tu_dai_nhat)
```

```python
nums = list(map(int, input("Nhập danh sách số: ").split()))
unique_nums = sorted(set(nums))
second_largest = unique_nums[-2]
print("Sorted list:", unique_nums)
print("Second largest:", second_largest)
```

```python
scores = list(map(float, input("Nhập danh sách điểm: ").split()))
valid_scores = []
for s in scores:
    if 0 <= s <= 10:
        valid_scores.append(s)
if len(valid_scores) == 0:
    print("Không có điểm hợp lệ")
else:
    average = sum(valid_scores) / len(valid_scores)
    passed = 0
    for s in valid_scores:
        if s >= 5:
            passed += 1
    print("Danh sách điểm hợp lệ:", valid_scores)
    print("Điểm trung bình:", round(average, 2))
    print("Số sinh viên đạt:", passed)
```
